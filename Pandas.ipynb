```python
import pandas as pd

# Criando um DataFrame
dados = {
    "nome": ["Ana", "Carlos", "Marcel", "João"],
    "idade": [25, 32, 28, 41],
    "salario": [3500, 5200, 4500, 7000]
}

df = pd.DataFrame(dados)

# Mostrar os dados
print(df)

# Média salarial
media = df["salario"].mean()

print(f"\nMédia salarial: R$ {media:.2f}")

# Pessoas com salário acima da média
print("\nSalários acima da média:")
print(df[df["salario"] > media])
```
