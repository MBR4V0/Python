```python
# Importando bibliotecas
import numpy as np
from sklearn.linear_model import LinearRegression

# ---------------------------------------
# Dados de treinamento (Queremos prever salário)
# ---------------------------------------

# Variáveis independentes:
# [experiência, horas_estudo, projetos]

X = np.array([
    [1, 2, 1],
    [2, 4, 2],
    [3, 5, 2],
    [4, 7, 3],
    [5, 8, 4],
    [6, 9, 5]
])

# Salários correspondentes
y = np.array([
    2000,
    3000,
    3500,
    5000,
    6500,
    8000
])

# ---------------------------------------
# Criando modelo
# ---------------------------------------

modelo = LinearRegression()

# Treinando modelo
modelo.fit(X, y)

# ---------------------------------------
# Fazendo previsão
# ---------------------------------------

# Novo funcionário:
# [4 anos experiência, 6 horas estudo, 3 projetos]

novo = np.array([[4, 6, 3]])

salario_previsto = modelo.predict(novo)

print(f"Salário previsto: R$ {salario_previsto[0]:.2f}")

# ---------------------------------------
# Coeficientes
# ---------------------------------------

print("\nIntercepto:")
print(modelo.intercept_)

print("\nCoeficientes:")

variaveis = [
    "Experiência",
    "Horas de estudo",
    "Projetos"
]

for nome, coef in zip(variaveis, modelo.coef_):
    print(f"{nome}: {coef:.2f}")
```
