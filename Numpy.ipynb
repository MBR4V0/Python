```python
import numpy as np

# Criando um array
notas = np.array([7.5, 8.0, 9.5, 6.5, 8.5])

# Cálculos
media = np.mean(notas)
maior = np.max(notas)
menor = np.min(notas)
desvio = np.std(notas)

print(f"Média: {media:.2f}")
print(f"Maior nota: {maior}")
print(f"Menor nota: {menor}")
print(f"Desvio padrão: {desvio:.2f}")
```
