```python
# Importando bibliotecas
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

# -----------------------------
# Dados de exemplo
# -----------------------------

# Horas estudadas (X)
X = np.array([1, 2, 3, 4, 5, 6]).reshape(-1, 1)

# Notas (Y)
y = np.array([52, 55, 61, 65, 72, 75])

# -----------------------------
# Criando o modelo
# -----------------------------

modelo = LinearRegression()

# Treinando o modelo
modelo.fit(X, y)

# -----------------------------
# Fazendo previsões
# -----------------------------

# Previsão para 7 horas estudadas
horas = np.array([[7]])

previsao = modelo.predict(horas)

print(f"Nota prevista para 7 horas: {previsao[0]:.2f}")

# -----------------------------
# Coeficientes da reta
# -----------------------------

print("\nEquação da reta:")

a = modelo.intercept_
b = modelo.coef_[0]

print(f"y = {a:.2f} + {b:.2f}x")

# -----------------------------
# Visualização gráfica
# -----------------------------

plt.scatter(X, y, color='blue', label='Dados reais')

plt.plot(X, modelo.predict(X),
         color='red',
         label='Reta de regressão')

plt.xlabel('Horas estudadas')
plt.ylabel('Nota')
plt.title('Regressão Linear Simples')

plt.legend()

plt.show()
```
