```python
%%writefile Conta.py
class Conta:
  def __init__(self, titular, numero, saldo):
    self.saldo=saldo
    self.numero = numero
    self.titular=titular
  @property
  def saldo(self):
    return self._saldo

  @saldo.setter
  def saldo(self,saldo):
    if saldo < 0:
      print("o saldo não pode ser negativo")
    else:
      self._saldo=saldo
```

```python
from Conta import Conta
ct = Conta("Fabio",123,-1)
print(ct.saldo)
```
