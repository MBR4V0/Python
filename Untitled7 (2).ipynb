```python
import numpy as np
from collections import Counter
import psutil
import os
def mode_using_counter(n_integers):
    """Gera números aleatórios e retorna o mais frequente"""
    process = psutil.Process(os.getpid()) # Obter o processo atual
    mem_before = process.memory_info().rss / 1024 ** 2 # Memória antes (MB)
    random_integers = np.random.randint(1, 100_000, n_integers)
    c = Counter(random_integers)
    result = c.most_common(1)[0][0]
    mem_after = process.memory_info().rss / 1024 ** 2 # Memória depois
    print(f" Memória antes: {mem_before:.2f} MB")
    print(f" Memória depois: {mem_after:.2f} MB")
    print(f" Diferença de memória: {mem_after - mem_before:.2f} MB")
    return result
mode_using_counter(100_000)
# O código mede o consumo de memória antes e depois de:
# Gerar um array grande com numpy
# Contar frequências com Counter
# Isso permite observar quanto de memória o algoritmo consome durante a execução.
# Acesso à memória é mais lento que CPU
# A CPU é extremamente rápida
# A memória RAM é mais lenta
# Quanto mais dados o programa manipula, mais tempo ele gasta acessando memória.
# ache da CPU (efeito importante)
# A CPU usa memória cache (L1, L2, L3), que é muito mais rápida que a RAM.
# Se os dados cabem no cache → execução rápida
# Se não cabem → precisa buscar na RAM → mais lento
# Ferramentas como:
# - cProfile ajudam a analisar o tempo de execução
# - mprofiler ajudam a visualizar o consumo de memória
# Ambos são importantes para entender e otimizar o desempenho de aplicações.
```
