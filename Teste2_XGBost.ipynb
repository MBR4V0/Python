**Random Forest**

```python
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score
```

```python
X = df[['Idade', 'Tumor_cm', 'Linfonodos', 'Ki67']]
```

```python
y = df['TN']
```

```python
print("X:")
display(X)

print("y:")
display(y)
```

**Separar treinamento e teste (70% dos dados para ensinar o modelo e 30% para testar.)**

```python
X_treino, X_teste, y_treino, y_teste = train_test_split(
    X,
    y,
    test_size=0.30,
    random_state=42
)
```

```python
print("Treinamento:", len(X_treino))
print("Teste:", len(X_teste))
```

**Criar o XGBost**

```python
!pip install -q xgboost
```

```python
from xgboost import XGBClassifier
```

```python
modelo_xgb = XGBClassifier(
    n_estimators=100,
    max_depth=3,
    learning_rate=0.1,
    random_state=42
)
```

```python
modelo_xgb.fit(
    X_treino,
    y_treino
)
```

**Fazendo as previsões**

```python
previsao_rf = modelo_xgb.predict(
    X_teste
)
```

```python
previsao_xgb = modelo_xgb.predict(X_teste)
```

```python
print("Real:")
print(y_teste.values)

print()

print("Previsão:")
print(previsao_rf)
```

```python
acuracia_xgb = accuracy_score(
    y_teste,
    previsao_xgb
)
```

```python
print(
    f"Acurácia XGBoost: {acuracia_xgb:.2%}"
)
```
