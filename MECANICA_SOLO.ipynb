```python
import math

class NBR6122FoundationDesign:

    def __init__(self, spt_profile):
        """
        spt_profile: lista de valores N-SPT por camada
        """
        self.spt = spt_profile
        self.FS = 2.0  # fator de segurança típico (NBR 6122 - simplificado)

    # -----------------------------
    # Média ponderada simplificada
    # -----------------------------
    def average_spt(self):
        return sum(self.spt) / len(self.spt)

    # -----------------------------
    # Classificação do solo (SPT)
    # -----------------------------
    def soil_class(self, N):
        if N < 4:
            return "Solo muito mole / muito fofo"
        elif N < 10:
            return "Solo mole / fofo"
        elif N < 20:
            return "Consistência média"
        elif N < 40:
            return "Solo rígido / denso"
        else:
            return "Solo muito rígido / muito denso"

    # -----------------------------
    # Resistência de ponta estimada
    # (correlação simplificada SPT)
    # -----------------------------
    def ultimate_bearing_capacity(self, N):
        # q_ult ≈ k * N (kPa)
        # faixa típica simplificada: 12 a 20 N
        k = 15
        return k * N

    # -----------------------------
    # Tensão admissível (NBR 6122)
    # σadm = σult / FS
    # -----------------------------
    def allowable_stress(self, N):
        q_ult = self.ultimate_bearing_capacity(N)
        return q_ult / self.FS

    # -----------------------------
    # Estimativa de recalque qualitativo
    # -----------------------------
    def settlement_risk(self, N):
        if N < 5:
            return "ALTO recalque"
        elif N < 15:
            return "Recalque moderado"
        else:
            return "Baixo recalque"

    # -----------------------------
    # Verificação de fundação rasa
    # -----------------------------
    def shallow_foundation_check(self, sigma_aplicada, N):

        sigma_adm = self.allowable_stress(N)

        if sigma_aplicada <= sigma_adm:
            return True, sigma_adm
        else:
            return False, sigma_adm

    # -----------------------------
    # Recomendação de fundação
    # -----------------------------
    def foundation_recommendation(self, sigma_aplicada):

        N_avg = self.average_spt()

        ok, sigma_adm = self.shallow_foundation_check(sigma_aplicada, N_avg)

        print("\n📊 RELATÓRIO NBR 6122 (SIMPLIFICADO)")
        print("=" * 60)

        print(f"Média SPT: {N_avg:.2f}")
        print("Classificação:", self.soil_class(N_avg))
        print("Risco de recalque:", self.settlement_risk(N_avg))

        print(f"\nTensão aplicada: {sigma_aplicada:.2f} kPa")
        print(f"Tensão admissível (σadm): {sigma_adm:.2f} kPa")

        print("\n🔎 VERIFICAÇÃO FUNDAÇÃO RASA:")
        if ok:
            print("✔ Viável conforme critério simplificado da NBR 6122")
            print("👉 Sugestão: sapatas isoladas ou radier")
        else:
            print("❌ NÃO viável para fundação rasa")
            print("👉 Sugestão: fundações profundas (estacas ou tubulões)")

        print("\n📌 OBS:")
        print("- Método simplificado baseado em SPT")
        print("- Deve ser validado com sondagem completa e projeto geotécnico")


# -----------------------------
# EXEMPLO DE USO
# -----------------------------
if __name__ == "__main__":

    # perfil SPT (camadas)
    spt_data = [3, 5, 7, 12, 18, 25]

    # tensão aplicada da estrutura (ex: prédio leve)
    sigma_aplicada = 120  # kPa

    design = NBR6122FoundationDesign(spt_data)
    design.foundation_recommendation(sigma_aplicada)
```
