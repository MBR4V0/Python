**Robô numérico.**

```python
import random

# Número de interações informado pelo usuário
interacoes = int(input("Digite o número de interações: "))

for i in range(interacoes):
    # Robô 1 escolhe um número real aleatório
    robo1 = random.uniform(0, 100)

    # Robô 2 escolhe outro número real aleatório
    robo2 = random.uniform(0, 100)

    # Soma dos números
    resultado = robo1 + robo2

    print(f"\nInteração {i + 1}")
    print(f"Robô 1: {robo1:.2f}")
    print(f"Robô 2: {robo2:.2f}")
    print(f"Resultado da soma: {resultado:.2f}")
```
