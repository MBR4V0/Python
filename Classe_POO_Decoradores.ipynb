```python
import pandas as pd

# -------------------------
# 1. CRIANDO O CSV
# -------------------------
df_vendas = pd.DataFrame({
    'Produto': ['Notebook', 'Mouse', 'Teclado', 'Notebook', 'Cadeira'],
    'Categoria': ['Eletrônicos', 'Periféricos', 'Periféricos', 'Eletrônicos', 'Geral'],
    'Quantidade': [2, 10, 5, 1, 3],
    'PrecoUnitario': [3500.00, 80.00, 150.00, 3600.00, 500.00]
})

df_vendas.to_csv('vendas.csv', index=False)
print("Arquivo 'vendas.csv' criado.")


# -------------------------
# 2. DECORADOR
# -------------------------
def valida_produto(func):
    def wrapper(self, produto):
        if not isinstance(produto, str) or produto.strip() == "":
            print("Produto inválido!")
            return None
        return func(self, produto)
    return wrapper


# -------------------------
# 3. CLASSE ANALISADOR
# -------------------------
class AnalisadorDeVendas:

    def __init__(self, caminho_csv):
        try:
            self.df = pd.read_csv(caminho_csv)
            print("Dataset carregado com sucesso!")
        except:
            print("Erro ao carregar o arquivo!")

    # -------------------------
    # dados (somente leitura)
    # -------------------------
    @property
    def dados(self):
        return self.df

    # -------------------------
    # faturamento total
    # -------------------------
    def faturamento_total(self):
        self.df['Faturamento'] = self.df['Quantidade'] * self.df['PrecoUnitario']
        return self.df['Faturamento'].sum()

    # -------------------------
    # resumo por categoria
    # -------------------------
    def resumo_por_categoria(self):
        self.df['Faturamento'] = self.df['Quantidade'] * self.df['PrecoUnitario']
        return self.df.groupby('Categoria')['Faturamento'].sum()

    # -------------------------
    # filtrar por produto (com decorador)
    # -------------------------
    @valida_produto
    def filtrar_por_produto(self, produto):
        return self.df[self.df['Produto'] == produto]

    # -------------------------
    # exportar resumo
    # -------------------------
    def exportar_resumo(self, caminho_saida):
        resumo = self.resumo_por_categoria()
        resumo.to_csv(caminho_saida)
        print("Resumo exportado com sucesso!")


# -------------------------
# 4. USO DA CLASSE
# -------------------------
analisador = AnalisadorDeVendas('vendas.csv')

print("\nFaturamento total:")
print(analisador.faturamento_total())

print("\nResumo por categoria:")
print(analisador.resumo_por_categoria())

print("\nFiltrar produto Notebook:")
print(analisador.filtrar_por_produto("Notebook"))

analisador.exportar_resumo("resumo.csv")
```
