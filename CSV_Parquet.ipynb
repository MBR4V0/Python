**Conversão de CSV para Parquet**

```python
import sys
from awsglue.transforms import *
from awsglue.utils import getResolvedOptions
from pyspark.context import SparkContext
from awsglue.context import GlueContext
from awsglue.job import Job

# Inicializa o contexto do Glue e do Spark
args = getResolvedOptions(sys.argv, ['JOB_NAME'])
sc = SparkContext()
glueContext = GlueContext(sc)
spark = glueContext.spark_session
job = Job(glueContext)
job.init(args['JOB_NAME'], args)

# 1. Lê os dados usando o Glue Catalog (Substitua pelos seus nomes)
datasource = glueContext.create_dynamic_frame.from_catalog(
    database = "meu_banco_de_dados",
    table_name = "minha_tabela_csv"
)

# 2. Grava os dados diretamente no S3 em formato Parquet
# O Glue cria a estrutura de pastas automaticamente
glueContext.write_dynamic_frame.from_options(
    frame = datasource,
    connection_type = "s3",
    connection_options = {"path": "s3://meu-bucket-de-destino/dados_parquet/"},
    format = "parquet"
)

job.commit()
```
