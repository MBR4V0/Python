```python
from itertools import combinations

# ============ 1) VARIÁVEIS DO PROBLEMA (edite aqui) ============
orcamento = 10  # R$ mil disponíveis

projetos = {
    # nome:                       (custo, utilidade)
    "A - Atualizar computadores": (6, 11),
    "B - Implantar backup":       (4, 8),
    "C - Melhorar a rede":        (5, 9),
    "D - Treinar a equipe":       (3, 7),
}

# ============ 2) ENUMERAR TODAS AS COMBINAÇÕES ============
nomes = list(projetos.keys())
resultados = []

for k in range(len(nomes) + 1):                 # tamanhos de 0 a 4 projetos
    for combo in combinations(nomes, k):
        custo_total = sum(projetos[p][0] for p in combo)
        utilidade_total = sum(projetos[p][1] for p in combo)
        viavel = custo_total <= orcamento       # restrição orçamentária
        resultados.append((combo, custo_total, utilidade_total, viavel))

# ============ 3) MOSTRAR TODAS AS COMBINAÇÕES ============
print(f"{'Combinação':<62}{'Custo':>6}{'Utilid.':>8}  Viável?")
for combo, c, u, v in resultados:
    texto = " + ".join(p[0] for p in combo) or "(nenhum)"
    print(f"{texto:<62}{c:>6}{u:>8}  {'sim' if v else 'não'}")

# ============ 4) ESCOLHER A MELHOR (maior utilidade entre as viáveis) ============
viaveis = [r for r in resultados if r[3]]
melhor = max(viaveis, key=lambda r: r[2])

print("\n=== SOLUÇÃO ÓTIMA ===")
for p in melhor[0]:
    print(" •", p)
print(f"Custo total: R$ {melhor[1]} mil (orçamento: R$ {orcamento} mil)")
print(f"Utilidade total: {melhor[2]}")
```

Etapas principais
Variáveis (bloco 1): é aqui que você altera os dados. orcamento guarda o limite de gastos, e o dicionário projetos guarda (custo, utilidade) de cada projeto. Para mudar o problema, basta editar esses valores ou adicionar linhas.
Enumeração (bloco 2): combinations gera todos os subconjuntos possíveis de projetos (2⁴ = 16). Para cada um, calculamos custo total e utilidade total (aditiva) e verificamos a restrição custo_total <= orcamento.
Tabela (bloco 3): imprime todas as combinações, o que ajuda a explicar o raciocínio no trabalho.
Decisão (bloco 4): filtra só as combinações viáveis e escolhe a de maior utilidade com max.
Resultado esperado

A melhor combinação é A + B, com custo de R$ 10 mil e utilidade 19. As alternativas viáveis ficam abaixo: A+D = 18, B+C = 17 e C+D = 16.

Observação: a força bruta funciona bem aqui, mas com muitos projetos (30 ou mais) fica lenta, porque cresce como 2ⁿ.
